# Locate the correct repository

```python
# DAY 22 — CELL 1
# PROJECT SETUP

from pathlib import Path
import os
import sys


NOTEBOOK_START_FOLDER = Path.cwd().resolve()


expected_existing_files = [
    Path("app") / "resilient_model_client.py",
    Path("app") / "prompt_contract.py",
    Path("prompts") / "finance_assistant_v2.json",
]


repository_candidates = [
    NOTEBOOK_START_FOLDER,
    *NOTEBOOK_START_FOLDER.parents,
]


projects_area = Path(
    "/var/www/filebrowser/.projects"
)

if projects_area.exists():
    repository_candidates.extend(
        projects_area.glob(
            "*/sap-finance-ai-assistant"
        )
    )


valid_repositories = []

for candidate in repository_candidates:
    try:
        candidate = candidate.resolve()

        if (
            candidate.exists()
            and (candidate / ".git").exists()
            and candidate not in valid_repositories
        ):
            valid_repositories.append(candidate)

    except (OSError, RuntimeError):
        continue


if not valid_repositories:
    raise FileNotFoundError(
        "The sap-finance-ai-assistant Git repository "
        "could not be located."
    )


def repository_score(repository_root):
    return sum(
        (repository_root / relative_path).exists()
        for relative_path in expected_existing_files
    )


PROJECT_ROOT = max(
    valid_repositories,
    key=repository_score,
)

GIT_ROOT = PROJECT_ROOT

os.chdir(PROJECT_ROOT)


if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(
        0,
        str(PROJECT_ROOT),
    )


for folder_name in [
    "app",
    "data",
    "artifacts",
    "tests",
    "notes",
    "notebooks",
]:
    (
        PROJECT_ROOT / folder_name
    ).mkdir(
        parents=True,
        exist_ok=True,
    )


print("Correct project root:")
print(PROJECT_ROOT)

print("\nPython version:")
print(sys.version.split()[0])

print(
    "\n✅ Day 22 project setup completed."
)
```

# Create the governed context builder

```python
# DAY 22 — CELL 3
# CREATE app/governed_context_builder.py

from pathlib import Path
import py_compile
import textwrap


module_source = textwrap.dedent(
    r'''
    """Access-aware context construction with citation lineage."""

    from __future__ import annotations

    import hashlib
    import json
    import re

    from dataclasses import dataclass
    from datetime import datetime, timezone
    from pathlib import Path
    from typing import Any, Iterable


    STOP_WORDS = {
        "a", "an", "and", "are", "as", "at", "be",
        "before", "by", "do", "does", "for", "from",
        "how", "in", "is", "it", "of", "on", "or",
        "please", "should", "that", "the", "this",
        "to", "what", "when", "who", "with",
    }


    TOKEN_ALIASES = {
        "approvals": "approval",
        "approved": "approval",
        "approver": "approval",
        "entries": "entry",
        "journals": "journal",
        "payments": "payment",
        "posted": "posting",
        "records": "record",
        "requirements": "requirement",
        "retained": "retention",
    }


    DEFAULT_INJECTION_MARKERS = (
        "ignore previous instructions",
        "ignore all previous instructions",
        "disregard previous instructions",
        "override authorization",
        "bypass access",
        "bypass security",
        "reveal restricted",
        "reveal the system prompt",
        "show hidden instructions",
    )


    def _utc_now() -> str:
        return datetime.now(
            timezone.utc
        ).isoformat()


    def _sha256_text(value: str) -> str:
        return hashlib.sha256(
            value.encode("utf-8")
        ).hexdigest()


    def _normalized_tokens(value: str) -> set[str]:
        words = re.findall(
            r"[a-z0-9]+",
            str(value).lower(),
        )

        return {
            TOKEN_ALIASES.get(word, word)
            for word in words
            if word not in STOP_WORDS
            and len(word) > 1
        }


    def _contains_injection(
        value: str,
        markers: Iterable[str],
    ) -> bool:
        normalized = " ".join(
            str(value).lower().split()
        )

        return any(
            marker in normalized
            for marker in markers
        )


    def _token_estimate(value: str) -> int:
        return max(
            1,
            len(
                re.findall(
                    r"\S+",
                    str(value),
                )
            ),
        )


    @dataclass(frozen=True)
    class GovernedContextRequest:
        question: str
        user_role: str
        company_code: str
        top_k: int = 3
        token_budget: int = 400

        def __post_init__(self) -> None:
            clean_question = str(
                self.question
            ).strip()

            clean_role = str(
                self.user_role
            ).strip().upper()

            clean_company = str(
                self.company_code
            ).strip().upper()

            if not clean_question:
                raise ValueError(
                    "question cannot be empty"
                )

            if not clean_role:
                raise ValueError(
                    "user_role cannot be empty"
                )

            if not clean_company:
                raise ValueError(
                    "company_code cannot be empty"
                )

            if not isinstance(self.top_k, int):
                raise TypeError(
                    "top_k must be an integer"
                )

            if not 1 <= self.top_k <= 10:
                raise ValueError(
                    "top_k must be between 1 and 10"
                )

            if not isinstance(
                self.token_budget,
                int,
            ):
                raise TypeError(
                    "token_budget must be an integer"
                )

            if not 25 <= self.token_budget <= 4000:
                raise ValueError(
                    "token_budget must be between 25 and 4000"
                )

            object.__setattr__(
                self,
                "question",
                clean_question,
            )

            object.__setattr__(
                self,
                "user_role",
                clean_role,
            )

            object.__setattr__(
                self,
                "company_code",
                clean_company,
            )


    class GovernedContextBuilder:
        def __init__(
            self,
            allowed_classifications=(
                "PUBLIC",
                "INTERNAL",
            ),
            injection_markers=(
                DEFAULT_INJECTION_MARKERS
            ),
        ) -> None:
            self.allowed_classifications = {
                str(item).upper()
                for item in allowed_classifications
            }

            self.injection_markers = tuple(
                str(item).lower()
                for item in injection_markers
            )


        def build(
            self,
            request: GovernedContextRequest,
            candidates: list[dict[str, Any]],
        ) -> dict[str, Any]:
            if _contains_injection(
                request.question,
                self.injection_markers,
            ):
                return self._blocked_package(
                    request=request,
                    reason="prompt_injection_in_question",
                )

            query_tokens = _normalized_tokens(
                request.question
            )

            eligible_candidates = []
            rejected_candidates = []
            seen_chunk_ids = set()

            for original_candidate in candidates:
                candidate = dict(
                    original_candidate
                )

                chunk_id = str(
                    candidate.get(
                        "chunk_id",
                        "",
                    )
                ).strip()

                rejection_reasons = []

                if not chunk_id:
                    rejection_reasons.append(
                        "missing_chunk_id"
                    )

                if chunk_id in seen_chunk_ids:
                    rejection_reasons.append(
                        "duplicate_chunk_id"
                    )

                seen_chunk_ids.add(chunk_id)

                status = str(
                    candidate.get(
                        "status",
                        "",
                    )
                ).upper()

                if status != "APPROVED":
                    rejection_reasons.append(
                        "content_not_approved"
                    )

                allowed_roles = {
                    str(role).upper()
                    for role in candidate.get(
                        "allowed_roles",
                        [],
                    )
                }

                if (
                    request.user_role
                    not in allowed_roles
                    and "*"
                    not in allowed_roles
                ):
                    rejection_reasons.append(
                        "role_not_authorized"
                    )

                company_codes = {
                    str(code).upper()
                    for code in candidate.get(
                        "company_codes",
                        [],
                    )
                }

                if (
                    request.company_code
                    not in company_codes
                    and "*"
                    not in company_codes
                ):
                    rejection_reasons.append(
                        "company_code_not_authorized"
                    )

                classification = str(
                    candidate.get(
                        "classification",
                        "",
                    )
                ).upper()

                if (
                    classification
                    not in self.allowed_classifications
                ):
                    rejection_reasons.append(
                        "classification_not_allowed"
                    )

                candidate_text = " ".join(
                    [
                        str(
                            candidate.get(
                                "title",
                                "",
                            )
                        ),
                        str(
                            candidate.get(
                                "content",
                                "",
                            )
                        ),
                        " ".join(
                            str(tag)
                            for tag in candidate.get(
                                "tags",
                                [],
                            )
                        ),
                    ]
                )

                if _contains_injection(
                    candidate_text,
                    self.injection_markers,
                ):
                    rejection_reasons.append(
                        "prompt_injection_in_content"
                    )

                candidate_tokens = (
                    _normalized_tokens(
                        candidate_text
                    )
                )

                overlap = (
                    query_tokens
                    & candidate_tokens
                )

                relevance_score = (
                    len(overlap)
                    / max(
                        len(query_tokens),
                        1,
                    )
                )

                if relevance_score <= 0:
                    rejection_reasons.append(
                        "no_relevance"
                    )

                if rejection_reasons:
                    rejected_candidates.append(
                        {
                            "chunk_id": (
                                chunk_id
                                or "<missing>"
                            ),
                            "reasons": sorted(
                                set(
                                    rejection_reasons
                                )
                            ),
                        }
                    )

                    continue

                candidate[
                    "_relevance_score"
                ] = round(
                    relevance_score,
                    6,
                )

                eligible_candidates.append(
                    candidate
                )

            eligible_candidates.sort(
                key=lambda item: (
                    -item[
                        "_relevance_score"
                    ],
                    item["chunk_id"],
                )
            )

            selected_chunks = []
            remaining_budget = (
                request.token_budget
            )

            for candidate in (
                eligible_candidates
            ):
                if (
                    len(selected_chunks)
                    >= request.top_k
                    or remaining_budget <= 0
                ):
                    break

                original_content = str(
                    candidate.get(
                        "content",
                        "",
                    )
                ).strip()

                words = (
                    original_content.split()
                )

                truncated = False

                if len(words) > remaining_budget:
                    if (
                        selected_chunks
                        or remaining_budget < 20
                    ):
                        rejected_candidates.append(
                            {
                                "chunk_id": candidate[
                                    "chunk_id"
                                ],
                                "reasons": [
                                    "token_budget_exceeded"
                                ],
                            }
                        )

                        continue

                    content = " ".join(
                        words[
                            :remaining_budget
                        ]
                    )

                    truncated = True

                else:
                    content = original_content

                token_estimate = (
                    _token_estimate(content)
                )

                if token_estimate > remaining_budget:
                    continue

                remaining_budget -= (
                    token_estimate
                )

                selected_chunks.append(
                    {
                        "chunk_id": candidate[
                            "chunk_id"
                        ],
                        "title": candidate.get(
                            "title",
                            "",
                        ),
                        "content": content,
                        "source_system": (
                            candidate.get(
                                "source_system",
                                "",
                            )
                        ),
                        "source_uri": (
                            candidate.get(
                                "source_uri",
                                "",
                            )
                        ),
                        "classification": (
                            candidate.get(
                                "classification",
                                "",
                            )
                        ),
                        "relevance_score": (
                            candidate[
                                "_relevance_score"
                            ]
                        ),
                        "token_estimate": (
                            token_estimate
                        ),
                        "truncated": truncated,
                        "content_sha256": (
                            _sha256_text(
                                original_content
                            )
                        ),
                    }
                )

            citations = [
                chunk["chunk_id"]
                for chunk in selected_chunks
            ]

            context_text = "\n\n".join(
                (
                    f"[{chunk['chunk_id']}] "
                    f"{chunk['title']}\n"
                    f"{chunk['content']}"
                )
                for chunk in selected_chunks
            )

            lineage = [
                {
                    "citation": chunk[
                        "chunk_id"
                    ],
                    "source_system": chunk[
                        "source_system"
                    ],
                    "source_uri": chunk[
                        "source_uri"
                    ],
                    "content_sha256": chunk[
                        "content_sha256"
                    ],
                }
                for chunk in selected_chunks
            ]

            package_identity = json.dumps(
                {
                    "question_hash": (
                        _sha256_text(
                            request.question
                        )
                    ),
                    "user_role": (
                        request.user_role
                    ),
                    "company_code": (
                        request.company_code
                    ),
                    "citations": citations,
                },
                sort_keys=True,
                separators=(",", ":"),
            )

            approved = bool(
                selected_chunks
            )

            return {
                "package_id": (
                    "CTX-"
                    + _sha256_text(
                        package_identity
                    )[:16].upper()
                ),
                "created_at_utc": _utc_now(),
                "request": {
                    "question_hash": (
                        _sha256_text(
                            request.question
                        )
                    ),
                    "user_role": (
                        request.user_role
                    ),
                    "company_code": (
                        request.company_code
                    ),
                    "top_k": request.top_k,
                    "token_budget": (
                        request.token_budget
                    ),
                },
                "governance": {
                    "status": (
                        "APPROVED"
                        if approved
                        else "BLOCKED"
                    ),
                    "approved_for_generation": (
                        approved
                    ),
                    "reason": (
                        "authorized_context_available"
                        if approved
                        else
                        "no_authorized_relevant_context"
                    ),
                },
                "citations": citations,
                "context_text": context_text,
                "chunks": selected_chunks,
                "lineage": lineage,
                "rejected_candidates": (
                    rejected_candidates
                ),
                "metrics": {
                    "candidate_count": len(
                        candidates
                    ),
                    "authorized_relevant_count": (
                        len(
                            eligible_candidates
                        )
                    ),
                    "selected_count": len(
                        selected_chunks
                    ),
                    "token_budget": (
                        request.token_budget
                    ),
                    "tokens_used": sum(
                        chunk[
                            "token_estimate"
                        ]
                        for chunk
                        in selected_chunks
                    ),
                },
            }


        def _blocked_package(
            self,
            request: GovernedContextRequest,
            reason: str,
        ) -> dict[str, Any]:
            return {
                "package_id": (
                    "CTX-BLOCKED-"
                    + _sha256_text(
                        request.question
                    )[:12].upper()
                ),
                "created_at_utc": _utc_now(),
                "request": {
                    "question_hash": (
                        _sha256_text(
                            request.question
                        )
                    ),
                    "user_role": (
                        request.user_role
                    ),
                    "company_code": (
                        request.company_code
                    ),
                    "top_k": request.top_k,
                    "token_budget": (
                        request.token_budget
                    ),
                },
                "governance": {
                    "status": "BLOCKED",
                    "approved_for_generation": (
                        False
                    ),
                    "reason": reason,
                },
                "citations": [],
                "context_text": "",
                "chunks": [],
                "lineage": [],
                "rejected_candidates": [],
                "metrics": {
                    "candidate_count": 0,
                    "authorized_relevant_count": 0,
                    "selected_count": 0,
                    "token_budget": (
                        request.token_budget
                    ),
                    "tokens_used": 0,
                },
            }


    def load_candidates(
        file_path: str | Path,
    ) -> list[dict[str, Any]]:
        payload = json.loads(
            Path(file_path).read_text(
                encoding="utf-8"
            )
        )

        if isinstance(payload, list):
            candidates = payload

        elif isinstance(payload, dict):
            candidates = payload.get(
                "candidates",
                [],
            )

        else:
            raise TypeError(
                "Candidate data must be a list "
                "or dictionary."
            )

        if not isinstance(candidates, list):
            raise TypeError(
                "The candidates field must be a list."
            )

        return candidates


    def build_context_package(
        question: str,
        user_role: str,
        company_code: str,
        candidates: list[dict[str, Any]],
        top_k: int = 3,
        token_budget: int = 400,
    ) -> dict[str, Any]:
        request = GovernedContextRequest(
            question=question,
            user_role=user_role,
            company_code=company_code,
            top_k=top_k,
            token_budget=token_budget,
        )

        builder = GovernedContextBuilder()

        return builder.build(
            request=request,
            candidates=candidates,
        )
    '''
).lstrip()


module_path = (
    PROJECT_ROOT
    / "app"
    / "governed_context_builder.py"
)

module_path.write_text(
    module_source,
    encoding="utf-8",
)


py_compile.compile(
    str(module_path),
    doraise=True,
)


print(
    "✅ Context builder created:",
    module_path.relative_to(PROJECT_ROOT),
)

print(
    "✅ Python compilation passed."
)
```

```python
# DAY 22 — RESTORE TEST VARIABLES

from pathlib import Path
import importlib.util
import sys


module_path = (
    PROJECT_ROOT
    / "app"
    / "governed_context_builder.py"
)


module_spec = (
    importlib.util.spec_from_file_location(
        "governed_context_builder",
        module_path,
    )
)


governed_context_builder = (
    importlib.util.module_from_spec(
        module_spec
    )
)


# Register the module before executing it.
sys.modules[
    module_spec.name
] = governed_context_builder


module_spec.loader.exec_module(
    governed_context_builder
)


candidate_path = (
    PROJECT_ROOT
    / "data"
    / "day22_retrieved_candidates.json"
)


candidates = (
    governed_context_builder
    .load_candidates(
        candidate_path
    )
)


context_builder = (
    governed_context_builder
    .GovernedContextBuilder()
)


print(
    "✅ Day 22 test variables restored."
)

print(
    "Candidates loaded:",
    len(candidates),
)
```

```python
# DAY 22 — CELL 5
# GOVERNED CONTEXT TESTS

from datetime import datetime, timezone
from pathlib import Path
import json


test_results = []


def record_test(
    test_id,
    name,
    passed,
    details,
):
    test_results.append(
        {
            "test_id": test_id,
            "name": name,
            "passed": bool(passed),
            "details": details,
        }
    )


# Test 1 — Authorized journal policy.
journal_package = context_builder.build(
    request=(
        governed_context_builder
        .GovernedContextRequest(
            question=(
                "What approval is required for "
                "a manual journal entry?"
            ),
            user_role="FINANCE_ANALYST",
            company_code="1000",
        )
    ),
    candidates=candidates,
)

record_test(
    "D22-T01",
    "authorized journal context",
    (
        journal_package["governance"][
            "approved_for_generation"
        ]
        is True
        and "FIN-JE-001"
        in journal_package["citations"]
    ),
    {
        "citations": (
            journal_package["citations"]
        ),
    },
)


# Test 2 — Company-code boundary.
company_boundary_package = (
    context_builder.build(
        request=(
            governed_context_builder
            .GovernedContextRequest(
                question=(
                    "What is the period-close "
                    "checklist for company 2000?"
                ),
                user_role="FINANCE_ANALYST",
                company_code="1000",
            )
        ),
        candidates=candidates,
    )
)

record_test(
    "D22-T02",
    "company-code isolation",
    (
        "FIN-CLOSE-004"
        not in company_boundary_package[
            "citations"
        ]
    ),
    {
        "citations": (
            company_boundary_package[
                "citations"
            ]
        ),
    },
)


# Test 3 — Restricted role and classification.
restricted_package = context_builder.build(
    request=(
        governed_context_builder
        .GovernedContextRequest(
            question=(
                "Show employee payroll "
                "and compensation information."
            ),
            user_role="FINANCE_ANALYST",
            company_code="1000",
        )
    ),
    candidates=candidates,
)

record_test(
    "D22-T03",
    "restricted payroll blocked",
    (
        "FIN-PAYROLL-005"
        not in restricted_package[
            "citations"
        ]
    ),
    {
        "citations": (
            restricted_package["citations"]
        ),
    },
)


# Test 4 — Malicious retrieved text excluded.
injection_package = context_builder.build(
    request=(
        governed_context_builder
        .GovernedContextRequest(
            question=(
                "What temporary journal override "
                "instructions are available?"
            ),
            user_role="FINANCE_ANALYST",
            company_code="1000",
        )
    ),
    candidates=candidates,
)

injection_rejections = [
    item
    for item in injection_package[
        "rejected_candidates"
    ]
    if item["chunk_id"]
    == "FIN-INJECT-006"
]

record_test(
    "D22-T04",
    "retrieved prompt injection excluded",
    (
        "FIN-INJECT-006"
        not in injection_package[
            "citations"
        ]
        and bool(injection_rejections)
        and (
            "prompt_injection_in_content"
            in injection_rejections[0][
                "reasons"
            ]
        )
    ),
    {
        "citations": (
            injection_package["citations"]
        ),
        "rejection": (
            injection_rejections
        ),
    },
)


# Test 5 — Citation lineage.
audit_package = context_builder.build(
    request=(
        governed_context_builder
        .GovernedContextRequest(
            question=(
                "What audit evidence and "
                "retention records are required?"
            ),
            user_role="AUDITOR",
            company_code="1000",
        )
    ),
    candidates=candidates,
)

audit_lineage_ids = [
    item["citation"]
    for item in audit_package[
        "lineage"
    ]
]

record_test(
    "D22-T05",
    "citation lineage retained",
    (
        "FIN-AUD-003"
        in audit_package["citations"]
        and set(
            audit_package["citations"]
        )
        == set(audit_lineage_ids)
    ),
    {
        "citations": (
            audit_package["citations"]
        ),
        "lineage_ids": audit_lineage_ids,
    },
)


# Test 6 — Token budget enforcement.
budget_package = context_builder.build(
    request=(
        governed_context_builder
        .GovernedContextRequest(
            question=(
                "journal payment audit close "
                "approval evidence posting"
            ),
            user_role="FINANCE_MANAGER",
            company_code="1000",
            top_k=5,
            token_budget=35,
        )
    ),
    candidates=candidates,
)

record_test(
    "D22-T06",
    "token budget enforced",
    (
        budget_package["metrics"][
            "tokens_used"
        ]
        <= 35
    ),
    {
        "tokens_used": (
            budget_package["metrics"][
                "tokens_used"
            ]
        ),
        "token_budget": 35,
    },
)


passed_tests = sum(
    result["passed"]
    for result in test_results
)

total_tests = len(test_results)


test_report = {
    "report_id": (
        "DAY22-GOVERNED-CONTEXT-TESTS"
    ),
    "generated_at_utc": datetime.now(
        timezone.utc
    ).isoformat(),
    "summary": {
        "passed_tests": passed_tests,
        "total_tests": total_tests,
        "pass_rate": (
            passed_tests / total_tests
            if total_tests
            else 0.0
        ),
    },
    "results": test_results,
}


test_report_path = (
    PROJECT_ROOT
    / "tests"
    / "day22_governed_context_results.json"
)

test_report_path.write_text(
    json.dumps(
        test_report,
        indent=2,
        sort_keys=True,
    )
    + "\n",
    encoding="utf-8",
)


for result in test_results:
    icon = (
        "✅"
        if result["passed"]
        else "❌"
    )

    print(
        icon,
        result["test_id"],
        result["name"],
    )


assert passed_tests == total_tests, (
    "One or more Day 22 governed-context "
    "tests failed."
)


print(
    f"\n✅ Day 22 tests passed: "
    f"{passed_tests}/{total_tests}"
)
```
